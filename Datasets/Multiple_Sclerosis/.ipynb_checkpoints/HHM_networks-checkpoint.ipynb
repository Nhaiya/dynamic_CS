```python
"""
HMM Network Construction for MS Dataset
========================================
Constructs brain networks using Hidden Markov Model (HMM) states instead of sliding windows.

This script:
1. Loads DATASET 2 MS patients (TR≈2.5) + matched Mendeley controls
2. Fits a group-level HMM with K=5 states on all subjects
3. Extracts state-specific connectivity for each subject
4. Saves networks as .mat files with shape (5, 116, 116)

Based on: Vidaurre et al. (2017) "Brain network dynamics are hierarchically organized in time"
"""

import pandas as pd
import numpy as np
import networkx as nx
from pathlib import Path
from scipy.io import savemat
import os
from hmmlearn.hmm import GaussianHMM
import warnings
warnings.filterwarnings('ignore')


def load_clinical_data(clinical_file='clinical.csv'):
    """Load clinical metadata for MS patients"""
    return pd.read_csv(clinical_file)


def load_timeseries_data(patient_id, visit=1, data_dir='AAL_timeseries'):
    """Load time series data from CSV file"""
    filename = f"AAL_timeseries_{patient_id}_{visit}.csv"
    filepath = Path(data_dir) / filename
    
    if not filepath.exists():
        raise FileNotFoundError(f"File {filename} not found in {data_dir}")
    
    data = pd.read_csv(filepath, header=None).values  # Shape: (116, timepoints)
    return data


def prepare_hmm_data(clinical_file='clinical.csv', data_dir='AAL_timeseries', 
                     dataset_filter=2, visits=[1, 2]):
    """
    Prepare data for HMM fitting.
    
    Loads DATASET 2 patients + matched controls.
    Returns list of subject dictionaries with time series.
    """
    clinical_data = load_clinical_data(clinical_file)
    
    # Filter to DATASET 2 only (TR≈2.5, matched with controls)
    dataset_2 = clinical_data[clinical_data['DATASET'] == dataset_filter]
    
    subjects = []
    
    print(f"Loading DATASET {dataset_filter} patients (TR≈2.5)...")
    
    for _, row in dataset_2.iterrows():
        patient_id = row['ID']
        
        for visit in visits:
            try:
                timeseries = load_timeseries_data(patient_id, visit, data_dir)
                # Transpose to (timepoints, 116) for HMM
                timeseries = timeseries.T
                
                subjects.append({
                    'subject_id': f"{patient_id}_visit{visit}",
                    'timeseries': timeseries,
                    'patient_id': patient_id,
                    'visit': visit,
                    'group': 'patient',
                    'dataset': dataset_filter,
                    'TR': 2.5 if dataset_filter == 2 else 2.2
                })
                
                print(f"  ✓ {patient_id} visit {visit}: shape {timeseries.shape}")
                
            except Exception as e:
                print(f"  ✗ {patient_id} visit {visit}: {e}")
    
    print(f"\nLoaded {len(subjects)} patient scans from DATASET {dataset_filter}")
    return subjects


def fit_group_hmm(subjects, n_states=5, n_iter=100):
    """
    Fit group-level HMM on all subjects.
    
    Parameters:
    -----------
    subjects : list
        List of subject dictionaries with 'timeseries' key
    n_states : int
        Number of brain states (K)
    n_iter : int
        Number of HMM iterations
    
    Returns:
    --------
    hmm : GaussianHMM
        Fitted HMM model
    """
    print("=" * 70)
    print(f"FITTING GROUP-LEVEL HMM (K={n_states} states)")
    print("=" * 70)
    
    # Stack all time series
    all_timeseries = []
    lengths = []
    
    for subj in subjects:
        ts = subj['timeseries']  # (timepoints, 116)
        all_timeseries.append(ts)
        lengths.append(len(ts))
    
    # Concatenate all time series
    X = np.vstack(all_timeseries)  # (total_timepoints, 116)
    
    print(f"Total subjects: {len(subjects)}")
    print(f"Total timepoints: {X.shape[0]}")
    print(f"Number of ROIs: {X.shape[1]}")
    print(f"Number of states: {n_states}")
    print(f"HMM iterations: {n_iter}")
    print("\nFitting HMM (this may take a few minutes)...")
    
    # Fit HMM
    hmm = GaussianHMM(
        n_components=n_states,
        covariance_type='full',
        n_iter=n_iter,
        random_state=42,
        verbose=False
    )
    
    hmm.fit(X, lengths)
    
    print("✓ HMM fitting complete!")
    print(f"  Final log-likelihood: {hmm.score(X, lengths):.2f}")
    print("=" * 70)
    
    return hmm


def extract_state_connectivity(subject_timeseries, hmm, n_states=5, threshold=0.4):
    """
    Extract state-specific connectivity matrices for one subject.
    
    Parameters:
    -----------
    subject_timeseries : np.ndarray
        Time series shape (timepoints, 116)
    hmm : GaussianHMM
        Fitted HMM model
    n_states : int
        Number of states
    threshold : float
        Correlation threshold
    
    Returns:
    --------
    state_networks : np.ndarray
        Shape (n_states, 116, 116) - connectivity per state
    state_counts : np.ndarray
        Number of timepoints per state
    """
    # Predict state sequence
    state_sequence = hmm.predict(subject_timeseries)
    
    state_networks = []
    state_counts = []
    
    for k in range(n_states):
        # Get timepoints belonging to state k
        state_mask = (state_sequence == k)
        state_timepoints = subject_timeseries[state_mask, :]
        
        n_timepoints = np.sum(state_mask)
        state_counts.append(n_timepoints)
        
        # Compute correlation if enough timepoints
        if n_timepoints >= 10:  # Minimum 10 timepoints for stable correlation
            corr_matrix = np.corrcoef(state_timepoints.T)  # (116, 116)
            corr_matrix = np.nan_to_num(corr_matrix, nan=0.0)
        else:
            # Fallback: use whole time series if state has too few timepoints
            corr_matrix = np.corrcoef(subject_timeseries.T)
            corr_matrix = np.nan_to_num(corr_matrix, nan=0.0)
        
        # Apply threshold
        edge_mask = (np.abs(corr_matrix) >= threshold)
        np.fill_diagonal(edge_mask, False)
        
        weighted_adjacency = corr_matrix.copy()
        weighted_adjacency[~edge_mask] = 0.0
        np.fill_diagonal(weighted_adjacency, 0)
        
        state_networks.append(weighted_adjacency)
    
    return np.array(state_networks), np.array(state_counts)


def process_all_subjects_hmm(subjects, hmm, n_states=5, threshold=0.4, 
                              output_dir='Networks_MS_HMM'):
    """
    Process all subjects and extract HMM state-specific networks.
    
    Parameters:
    -----------
    subjects : list
        List of subject dictionaries
    hmm : GaussianHMM
        Fitted HMM model
    n_states : int
        Number of states
    threshold : float
        Correlation threshold
    output_dir : str
        Output directory for .mat files
    """
    os.makedirs(output_dir, exist_ok=True)
    
    print("\n" + "=" * 70)
    print("EXTRACTING STATE-SPECIFIC NETWORKS")
    print("=" * 70)
    print(f"Processing {len(subjects)} subjects")
    print(f"Number of states: {n_states}")
    print(f"Threshold: {threshold}")
    print(f"Output directory: {output_dir}")
    print("=" * 70)
    
    success_count = 0
    fail_count = 0
    
    for i, subject in enumerate(subjects):
        subject_id = subject['subject_id']
        
        try:
            # Extract state-specific connectivity
            state_networks, state_counts = extract_state_connectivity(
                subject['timeseries'],
                hmm,
                n_states=n_states,
                threshold=threshold
            )
            
            # Save to .mat file
            matlab_filename = f"{subject_id}_hmm.mat"
            matlab_filepath = os.path.join(output_dir, matlab_filename)
            
            matlab_data = {
                'weighted_adjacency_matrices': state_networks,  # (K, 116, 116)
                'state_counts': state_counts,  # (K,) - timepoints per state
                'n_states': n_states,
                'threshold': threshold,
                'patient_id': subject['patient_id'],
                'visit': subject['visit'],
                'group': subject['group'],
                'dataset': subject['dataset'],
                'TR': subject['TR'],
                'method': 'HMM'
            }
            
            savemat(matlab_filepath, matlab_data)
            
            success_count += 1
            state_dist = ", ".join([f"S{k}:{state_counts[k]}" for k in range(n_states)])
            print(f"✓ [{i+1}/{len(subjects)}]: {subject_id} - {subject['group']} "
                  f"({state_dist} timepoints)")
            
        except Exception as e:
            fail_count += 1
            print(f"✗ [{i+1}/{len(subjects)}]: {subject_id} - {e}")
    
    print("=" * 70)
    print(f"COMPLETE: {success_count} successful, {fail_count} failed")
    print(f"Networks saved to: {output_dir}/")
    print("=" * 70)
    
    return success_count, fail_count


# ==============================================================================
# MAIN EXECUTION
# ==============================================================================

if __name__ == "__main__":
    
    # Configuration
    clinical_file = 'clinical.csv'
    data_dir = 'AAL_timeseries'
    n_states = 5  # Number of HMM states
    threshold = 0.4
    output_dir = 'Networks_MS_HMM'
    
    print("=" * 70)
    print("HMM NETWORK CONSTRUCTION - MS DATASET 2 + MATCHED CONTROLS")
    print("=" * 70)
    print(f"Configuration:")
    print(f"  - Number of states (K): {n_states}")
    print(f"  - Threshold: {threshold}")
    print(f"  - Dataset: DATASET 2 (TR≈2.5)")
    print(f"  - Visits: 1 and 2")
    print("=" * 70)
    
    # Step 1: Load DATASET 2 patients
    print("\n" + "="*70)
    print("STEP 1: LOADING DATA")
    print("="*70)
    
    subjects = prepare_hmm_data(
        clinical_file=clinical_file,
        data_dir=data_dir,
        dataset_filter=2,  # Only DATASET 2
        visits=[1, 2]
    )
    
    if len(subjects) == 0:
        print("✗ No subjects loaded. Exiting.")
        exit(1)
    
    # Step 2: Fit group-level HMM
    print("\n" + "="*70)
    print("STEP 2: FITTING HMM")
    print("="*70)
    
    hmm = fit_group_hmm(subjects, n_states=n_states, n_iter=100)
    
    # Step 3: Extract state-specific networks
    print("\n" + "="*70)
    print("STEP 3: EXTRACTING NETWORKS")
    print("="*70)
    
    success, fail = process_all_subjects_hmm(
        subjects,
        hmm,
        n_states=n_states,
        threshold=threshold,
        output_dir=output_dir
    )
    
    print("\n" + "="*70)
    print("✓ HMM NETWORK CONSTRUCTION COMPLETE!")
    print("="*70)
    print(f"Processed {len(subjects)} subjects")
    print(f"Output: {output_dir}/")
    print(f"Network shape per subject: ({n_states}, 116, 116)")
    print("="*70)
```
