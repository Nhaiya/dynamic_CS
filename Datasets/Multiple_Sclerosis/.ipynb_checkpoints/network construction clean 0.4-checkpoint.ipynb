# <center>Dynamic Multilayer Brain Network Construction</center>

```python
import pandas as pd
import numpy as np
import networkx as nx
from pathlib import Path
from scipy.io import savemat
import os 

def load_clinical_data(clinical_file='clinical.csv'):
    return pd.read_csv(clinical_file)

def get_patient_info(patient_id, clinical_data):
    patient_row = clinical_data[clinical_data['ID'] == patient_id]
    if patient_row.empty:
        raise ValueError(f"Patient {patient_id} not found in clinical data")
    
    dataset = patient_row['DATASET'].iloc[0]
    therapy = patient_row['THERAPY'].iloc[0]
    
    if dataset == 1:
        TR = 2.2
        total_volumes_expected = 300
    else:
        TR = 2.5
        total_volumes_expected = 240
    
    return {
        'dataset': dataset,
        'therapy': therapy,
        'TR': TR,
        'expected_volumes': total_volumes_expected,
        'patient_info': patient_row.iloc[0].to_dict()
    }

def load_timeseries_data(patient_id, visit=1, data_dir='.'):
    filename = f"AAL_timeseries_{patient_id}_{visit}.csv"
    filepath = Path(data_dir) / filename
    
    if not filepath.exists():
        raise FileNotFoundError(f"File {filename} not found in {data_dir}")
    
    data = pd.read_csv(filepath, header=None).values
    
    return data

def calculate_time_windows(n_timepoints, TR, window_duration=30):
    window_size = int(window_duration / TR)
    n_windows = n_timepoints // window_size
    
    return window_size, n_windows

def create_time_windows(data, window_size, n_windows):
    windows = []
    for i in range(n_windows):
        start_idx = i * window_size
        end_idx = start_idx + window_size
        window_data = data[:, start_idx:end_idx]
        windows.append(window_data)
    
    return windows

def calculate_correlation_matrix(window_data):
    correlation_matrix = np.corrcoef(window_data)
    correlation_matrix = np.nan_to_num(correlation_matrix, nan=0.0)
    
    return correlation_matrix

def create_weighted_network_layer(correlation_matrix, threshold):
    edge_mask = (np.abs(correlation_matrix) >= threshold)
    np.fill_diagonal(edge_mask, False)
    
    weighted_adjacency = correlation_matrix.copy()
    weighted_adjacency[~edge_mask] = 0.0
    np.fill_diagonal(weighted_adjacency, 0)
    
    G = nx.from_numpy_array(weighted_adjacency)
    
    for i, j in G.edges():
        G[i][j]['correlation'] = correlation_matrix[i, j]
        G[i][j]['abs_correlation'] = abs(correlation_matrix[i, j])
        
    return G, weighted_adjacency

def construct_multilayer_network(patient_id, visit=1, window_duration=30,
                                 absolute_threshold=0.4,
                                 clinical_file='clinical.csv', data_dir='.'):
    
    clinical_data = load_clinical_data(clinical_file)
    patient_info = get_patient_info(patient_id, clinical_data)
    
    timeseries_data = load_timeseries_data(patient_id, visit, data_dir)
    
    n_regions, n_timepoints = timeseries_data.shape
    
    window_size, n_windows = calculate_time_windows(
        n_timepoints, patient_info['TR'], window_duration
    )
    
    if n_windows == 0:
        raise ValueError(f"No complete windows possible with current parameters")
    
    windows = create_time_windows(timeseries_data, window_size, n_windows)
    
    correlation_matrices = []
    for i, window_data in enumerate(windows):
        corr_matrix = calculate_correlation_matrix(window_data)
        correlation_matrices.append(corr_matrix)
    
    layers = []
    weighted_adjacency_matrices = []

    for i, corr_matrix in enumerate(correlation_matrices):
        layer_graph, weighted_adj_matrix = create_weighted_network_layer(corr_matrix, absolute_threshold)
        layers.append(layer_graph)
        weighted_adjacency_matrices.append(weighted_adj_matrix)
    
    results = {
        'layers': layers,
        'weighted_adjacency_matrices': weighted_adjacency_matrices,
    }

    networks_dir = "Networks_test"
    os.makedirs(networks_dir, exist_ok=True)
    matlab_filename = f"{patient_id}_visit{visit}.mat"
    matlab_filepath = os.path.join(networks_dir, matlab_filename)

    matlab_data = {
        'weighted_adjacency_matrices': np.array(weighted_adjacency_matrices)
    }
    
    savemat(matlab_filepath, matlab_data)  
    
    return results

# Example usage
if __name__ == "__main__":
    
    clinical_csv_path = 'clinical.csv' 
    data_dir = 'AAL_timeseries'

    patient_ids_to_test = ['p001'] 
    visits_to_test = [1]
    
    absolute_thresholds = [0.4]
    
    window_duration_sec = 30

    all_results = {}

    for p_id in patient_ids_to_test:
        for v_id in visits_to_test:
            patient_results_for_visit = {}
            for threshold in absolute_thresholds:
                try:
                    results = construct_multilayer_network(
                        patient_id=p_id,
                        visit=v_id,
                        window_duration=window_duration_sec,
                        absolute_threshold=threshold,
                        clinical_file=clinical_csv_path,
                        data_dir=data_dir
                    )
                    if results: 
                        patient_results_for_visit[f'threshold_{int(threshold*100)}'] = results

                except FileNotFoundError as e:
                    print(f"Skipping Patient {p_id}, Visit {v_id}, Threshold {threshold:.2f}: {e}")
                except ValueError as e:
                    print(f"Skipping Patient {p_id}, Visit {v_id}, Threshold {threshold:.2f} due to data format issue: {e}")
                except Exception as e:
                    print(f"An unexpected error occurred for Patient {p_id}, Visit {v_id}, Threshold {threshold:.2f}: {e}")
            all_results[f'patient_{p_id}_visit_{v_id}'] = patient_results_for_visit
```
