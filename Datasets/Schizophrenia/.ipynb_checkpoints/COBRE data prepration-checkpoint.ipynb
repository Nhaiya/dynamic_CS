```python
import nibabel as nib
import numpy as np
import os
import glob
import pandas as pd
from nilearn.maskers import NiftiLabelsMasker

extract_dir = "/Users/noha/Downloads/aal_for_SPM8"
aal_map = os.path.join(extract_dir, "ROI_MNI_V4.nii")
aal_labels_file = os.path.join(extract_dir, "ROI_MNI_V4.txt")

aal_labels = []
with open(aal_labels_file, 'r') as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 2:
            aal_labels.append(parts[1])

atlas_img = nib.load(aal_map)
print(f"Atlas shape: {atlas_img.shape}")
print(f"Number of region labels: {len(aal_labels)}")

assert len(aal_labels) == 116, f"Expected 116 regions, got {len(aal_labels)}"
print("✓ Confirmed: This is the 116-region AAL atlas!")

masker = NiftiLabelsMasker(
    labels_img=aal_map,
    standardize=True,
    detrend=True,
    low_pass=0.1,
    high_pass=0.008,
    t_r=2.0,
    verbose=1
)

data_dir = "/Users/noha/Desktop/W2025/Data/COBRE"  # UPDATE THIS PATH!

demo_df = pd.read_csv(os.path.join(data_dir, "cobre_model_group.csv"))
demo_df.columns = demo_df.columns.str.strip()
demo_df['participant_id'] = demo_df.iloc[:, 0].str.strip()

cobre_files = glob.glob(os.path.join(data_dir, "fmri_*.nii.gz"))
print(f"\nFound {len(cobre_files)} COBRE fMRI files")

control_data = []
for i, subject_file in enumerate(cobre_files):
    filename = os.path.basename(subject_file)
    subject_id = filename.split('_')[1]
    
    print(f"\nProcessing {i+1}/{len(cobre_files)}: {subject_id}")
    
    try:
        subject_row = demo_df[demo_df['participant_id'] == subject_id]
        
        if len(subject_row) == 0:
            print(f"✗ No demographics for {subject_id}")
            continue
        
        age = int(subject_row['age'].values[0])
        gender = 'M' if subject_row['sex'].values[0] == 1 else 'F'
        diagnosis = 'Patient' if subject_row['sz'].values[0] == 1 else 'Control'
        
        img = nib.load(subject_file)
        time_series = masker.fit_transform(img)
        
        if time_series.shape[1] != 116:
            print(f"✗ Got {time_series.shape[1]} regions, skipping")
            continue
        
        control_data.append({
            'subject_id': subject_id,
            'time_series': time_series,
            'age': age,
            'gender': gender,
            'diagnosis': diagnosis
        })
        
        print(f"✓ {diagnosis} {gender}{age}, shape: {time_series.shape}")
        
    except Exception as e:
        print(f"✗ Error: {e}")
        continue

print(f"\n{'='*50}")
print(f"✓ Successfully processed {len(control_data)} subjects")
controls = [s for s in control_data if s['diagnosis'] == 'Control']
patients = [s for s in control_data if s['diagnosis'] == 'Patient']
print(f"Controls: {len(controls)}, Patients: {len(patients)}")
if control_data:
    ages = [s['age'] for s in control_data]
    print(f"Age range: {min(ages)} - {max(ages)}")
```

```python
import pandas as pd
import numpy as np
import networkx as nx
from pathlib import Path
from scipy.io import savemat
import os 

def calculate_time_windows(n_timepoints, TR=2, window_duration=30):
    """Calculate window size and number of windows"""
    window_size = int(window_duration / TR)
    n_windows = n_timepoints // window_size
    
    return window_size, n_windows

def create_time_windows(data, window_size, n_windows):
    """Split time series data into windows"""
    windows = []
    for i in range(n_windows):
        start_idx = i * window_size
        end_idx = start_idx + window_size
        window_data = data[start_idx:end_idx, :]
        windows.append(window_data)
    
    return windows

def calculate_correlation_matrix(window_data):
    """Calculate correlation matrix for a time window"""
    correlation_matrix = np.corrcoef(window_data.T)
    correlation_matrix = np.nan_to_num(correlation_matrix, nan=0.0)
    
    return correlation_matrix

def create_weighted_network_layer(correlation_matrix, threshold):
    """Create weighted network from correlation matrix"""
    edge_mask = (np.abs(correlation_matrix) >= threshold)
    np.fill_diagonal(edge_mask, False)
    
    weighted_adjacency = correlation_matrix.copy()
    weighted_adjacency[~edge_mask] = 0.0
    np.fill_diagonal(weighted_adjacency, 0)
    
    G = nx.from_numpy_array(weighted_adjacency)
    
    for i, j in G.edges():
        G[i][j]['correlation'] = correlation_matrix[i, j]
        G[i][j]['abs_correlation'] = abs(correlation_matrix[i, j])
        
    return G, weighted_adjacency

def construct_multilayer_network_control(subject_data, window_duration=30, 
                                        absolute_threshold=0.4, TR=2.0):
    
    timeseries_data = subject_data['time_series']
    subject_id = subject_data['subject_id'].replace('.nii.gz', '')
    
    n_timepoints, n_regions = timeseries_data.shape
    
    window_size, n_windows = calculate_time_windows(n_timepoints, TR, window_duration)
    
    if n_windows < 8:
        raise ValueError(f"Only {n_windows} windows, need at least 8")
    
    # Take first 8 windows only
    windows = create_time_windows(timeseries_data, window_size, 8)
    
    correlation_matrices = []
    for window_data in windows:
        corr_matrix = calculate_correlation_matrix(window_data)
        correlation_matrices.append(corr_matrix)
    
    layers = []
    weighted_adjacency_matrices = []
    
    for corr_matrix in correlation_matrices:
        layer_graph, weighted_adj_matrix = create_weighted_network_layer(
            corr_matrix, absolute_threshold
        )
        layers.append(layer_graph)
        weighted_adjacency_matrices.append(weighted_adj_matrix)
    
    results = {
        'layers': layers,
        'weighted_adjacency_matrices': weighted_adjacency_matrices,
        'subject_id': subject_id,
        'age': subject_data['age'],
        'gender': subject_data['gender'],
        'n_windows': 8,
        'n_regions': n_regions
    }
    
    return results

def process_all_controls(control_data, window_duration=30, absolute_threshold=0.4, 
                        TR=2.0, output_dir="Networks_controls"):
    
    os.makedirs(output_dir, exist_ok=True)
    
    all_results = []
    success_count = 0
    fail_count = 0
    
    print(f"Processing {len(control_data)} control subjects")
    print(f"Parameters: TR={TR}s, Window={window_duration}s, Threshold={absolute_threshold}")
    print("="*60)
    
    for i, subject_data in enumerate(control_data):
        subject_id = subject_data['subject_id'].replace('.nii.gz', '')
        
        try:
            results = construct_multilayer_network_control(
                subject_data,
                window_duration=window_duration,
                absolute_threshold=absolute_threshold,
                TR=TR
            )
            
            matlab_filename = f"{subject_id}.mat"
            matlab_filepath = os.path.join(output_dir, matlab_filename)
            
            matlab_data = {
                'weighted_adjacency_matrices': np.array(results['weighted_adjacency_matrices']),
                'age': results['age'],
                'gender': results['gender'],
                'n_windows': results['n_windows'],
                'n_regions': results['n_regions']
            }
            
            savemat(matlab_filepath, matlab_data)
            
            all_results.append(results)
            success_count += 1
            print(f"✓ SUCCESS [{i+1}/{len(control_data)}]: {subject_id} - "
                  f"{results['gender']}{results['age']}, {results['n_windows']} windows")
            
        except Exception as e:
            fail_count += 1
            print(f"✗ ERROR [{i+1}/{len(control_data)}]: {subject_id} - {e}")
    
    print("="*60)
    print(f"COMPLETE: {success_count} successful, {fail_count} failed")
    print(f"MATLAB files saved to: {output_dir}/")
    
    summary_data = []
    for result in all_results:
        summary_data.append({
            'subject_id': result['subject_id'],
            'age': result['age'],
            'gender': result['gender'],
            'n_windows': result['n_windows'],
            'n_regions': result['n_regions']
        })
    
    if summary_data:
        summary_df = pd.DataFrame(summary_data)
        summary_path = os.path.join(output_dir, 'processing_summary.csv')
        summary_df.to_csv(summary_path, index=False)
        print(f"Summary saved to: {summary_path}")
    
    return all_results

if __name__ == "__main__":
    results = process_all_controls(
        control_data,
        window_duration=30,
        absolute_threshold=0.4,
        TR=2,
        output_dir="COBRE_Networks"
    )
    
    print("\nProcessing complete!")
```

```python
"""
NEW CELL FOR COBRE NOTEBOOK: Std-Weighted Network Construction
===============================================================
Add this as a new cell in your COBRE_data_preparation.ipynb
after the existing cells that create control_data.

This cell constructs std-weighted networks for COBRE subjects (both controls and patients).
Takes first 8 windows only to match your existing COBRE processing.
"""

# ============================================================================
# STD-WEIGHTED NETWORK CONSTRUCTION FUNCTIONS FOR COBRE
# ============================================================================
import pandas as pd
import numpy as np
import networkx as nx
from pathlib import Path
from scipy.io import savemat
import os 


def calculate_time_windows(n_timepoints, TR=2, window_duration=30):
    """Calculate window size and number of windows"""
    window_size = int(window_duration / TR)
    n_windows = n_timepoints // window_size
    return window_size, n_windows


def create_time_windows(data, window_size, n_windows):
    """Split time series data into windows"""
    windows = []
    for i in range(n_windows):
        start_idx = i * window_size
        end_idx = start_idx + window_size
        window_data = data[start_idx:end_idx, :]
        windows.append(window_data)
    return windows


def calculate_std_weighted_correlation(window_data):
    """
    Calculate std-weighted correlation matrix.
    
    Edge weight = correlation[i,j] * sqrt(std[i] * std[j])
    
    Parameters:
    -----------
    window_data : np.ndarray
        Shape (window_size timepoints, 116 ROIs)
    
    Returns:
    --------
    np.ndarray : Std-weighted correlation matrix, shape (116, 116)
    """
    # Compute Pearson correlation
    correlation_matrix = np.corrcoef(window_data.T)
    correlation_matrix = np.nan_to_num(correlation_matrix, nan=0.0)
    
    # Compute standard deviation per ROI across time
    std_vector = np.std(window_data, axis=0)  # (116,)
    
    # Create std weight matrix: sqrt(std[i] * std[j])
    std_weight_matrix = np.sqrt(np.outer(std_vector, std_vector))
    
    # Apply std weighting
    std_weighted_matrix = correlation_matrix * std_weight_matrix
    
    # Handle any NaN or Inf values
    std_weighted_matrix = np.nan_to_num(std_weighted_matrix, nan=0.0, posinf=0.0, neginf=0.0)
    
    return std_weighted_matrix


def create_weighted_network_layer(std_weighted_matrix, threshold):
    """Apply threshold and create network layer"""
    edge_mask = (np.abs(std_weighted_matrix) >= threshold)
    np.fill_diagonal(edge_mask, False)
    
    weighted_adjacency = std_weighted_matrix.copy()
    weighted_adjacency[~edge_mask] = 0.0
    np.fill_diagonal(weighted_adjacency, 0)
    
    G = nx.from_numpy_array(weighted_adjacency)
    
    for i, j in G.edges():
        G[i][j]['weight'] = std_weighted_matrix[i, j]
        G[i][j]['abs_weight'] = abs(std_weighted_matrix[i, j])
    
    return G, weighted_adjacency


def construct_multilayer_network_std_weighted_cobre(subject_data, window_duration=30, 
                                                     absolute_threshold=0.4, TR=2.0):
    """
    Construct std-weighted multilayer network for COBRE subject.
    Takes first 8 windows only.
    """
    timeseries_data = subject_data['time_series']
    subject_id = subject_data['subject_id'].replace('.nii.gz', '')
    
    n_timepoints, n_regions = timeseries_data.shape
    
    # Calculate windows
    window_size, n_windows = calculate_time_windows(n_timepoints, TR, window_duration)
    
    if n_windows < 8:
        raise ValueError(f"Only {n_windows} windows, need at least 8")
    
    # Take first 8 windows only
    windows = create_time_windows(timeseries_data, window_size, 8)
    
    # Calculate std-weighted correlation for each window
    std_weighted_matrices = []
    for window_data in windows:
        std_weighted_matrix = calculate_std_weighted_correlation(window_data)
        std_weighted_matrices.append(std_weighted_matrix)
    
    # Create network layers with threshold
    layers = []
    weighted_adjacency_matrices = []
    
    for std_weighted_matrix in std_weighted_matrices:
        layer_graph, weighted_adj_matrix = create_weighted_network_layer(
            std_weighted_matrix, absolute_threshold
        )
        layers.append(layer_graph)
        weighted_adjacency_matrices.append(weighted_adj_matrix)
    
    results = {
        'layers': layers,
        'weighted_adjacency_matrices': weighted_adjacency_matrices,
        'subject_id': subject_id,
        'age': subject_data['age'],
        'gender': subject_data['gender'],
        'diagnosis': subject_data['diagnosis'],
        'n_windows': 8,
        'n_regions': n_regions
    }
    
    return results


def process_all_cobre_std_weighted(control_data, window_duration=30, 
                                    absolute_threshold=0.4, TR=2.0, 
                                    output_dir="COBRE_Networks_std_weighted"):
    """
    Process all COBRE subjects (controls + patients) with std-weighted correlation.
    Takes first 8 windows only.
    """
    os.makedirs(output_dir, exist_ok=True)
    
    all_results = []
    success_count = 0
    fail_count = 0
    
    print("=" * 70)
    print("STD-WEIGHTED NETWORK CONSTRUCTION - COBRE DATASET")
    print("=" * 70)
    print(f"Processing {len(control_data)} subjects")
    print(f"Parameters: TR={TR}s, Window={window_duration}s, Threshold={absolute_threshold}")
    print(f"Taking first 8 windows only")
    print(f"Output directory: {output_dir}")
    print("=" * 70)
    
    for i, subject_data in enumerate(control_data):
        subject_id = subject_data['subject_id'].replace('.nii.gz', '')
        
        try:
            # Construct std-weighted multilayer network
            results = construct_multilayer_network_std_weighted_cobre(
                subject_data,
                window_duration=window_duration,
                absolute_threshold=absolute_threshold,
                TR=TR
            )
            
            # Save to MATLAB format
            matlab_filename = f"{subject_id}_std_weighted.mat"
            matlab_filepath = os.path.join(output_dir, matlab_filename)
            
            matlab_data = {
                'weighted_adjacency_matrices': np.array(results['weighted_adjacency_matrices']),
                'age': results['age'],
                'gender': results['gender'],
                'diagnosis': results['diagnosis'],
                'n_windows': results['n_windows'],
                'n_regions': results['n_regions'],
                'threshold': absolute_threshold,
                'TR': TR,
                'method': 'std_weighted'
            }
            
            savemat(matlab_filepath, matlab_data)
            
            all_results.append(results)
            success_count += 1
            print(f"✓ SUCCESS [{i+1}/{len(control_data)}]: {subject_id} - "
                  f"{results['diagnosis']} {results['gender']}{results['age']}, "
                  f"{results['n_windows']} windows")
            
        except Exception as e:
            fail_count += 1
            print(f"✗ ERROR [{i+1}/{len(control_data)}]: {subject_id} - {e}")
    
    print("=" * 70)
    print(f"COMPLETE: {success_count} successful, {fail_count} failed")
    print(f"MATLAB files saved to: {output_dir}/")
    
    # Create summary CSV
    summary_data = []
    for result in all_results:
        summary_data.append({
            'subject_id': result['subject_id'],
            'age': result['age'],
            'gender': result['gender'],
            'diagnosis': result['diagnosis'],
            'n_windows': result['n_windows'],
            'n_regions': result['n_regions']
        })
    
    if summary_data:
        summary_df = pd.DataFrame(summary_data)
        summary_path = os.path.join(output_dir, 'processing_summary_std_weighted.csv')
        summary_df.to_csv(summary_path, index=False)
        print(f"Summary saved to: {summary_path}")
        
        # Print diagnosis breakdown
        controls = summary_df[summary_df['diagnosis'] == 'Control']
        patients = summary_df[summary_df['diagnosis'] == 'Patient']
        print(f"\nDiagnosis breakdown:")
        print(f"  Controls: {len(controls)}")
        print(f"  Patients: {len(patients)}")
        
    print("=" * 70)
    
    return all_results


# ============================================================================
# RUN STD-WEIGHTED PROCESSING FOR COBRE
# ============================================================================

# Process all COBRE subjects with std-weighted correlation
# Note: This assumes 'control_data' is already loaded from previous cells
std_weighted_results = process_all_cobre_std_weighted(
    control_data,
    window_duration=30,
    absolute_threshold=0.4,
    TR=2.0,
    output_dir="COBRE_Networks_std_weighted"
)

print("\n✓ Std-weighted network construction complete!")
print(f"✓ Processed {len(std_weighted_results)} subjects")
print(f"✓ Output saved to: COBRE_Networks_std_weighted/")
```

```python
"""
Organize COBRE networks into Controls and Patients folders
"""
import os
import shutil
from scipy.io import loadmat
from pathlib import Path

# Configuration
input_dir = "COBRE_Networks_std_weighted"
controls_dir = os.path.join(input_dir, "Controls")
patients_dir = os.path.join(input_dir, "Patients")

# Create subdirectories
os.makedirs(controls_dir, exist_ok=True)
os.makedirs(patients_dir, exist_ok=True)

# Get all .mat files
mat_files = list(Path(input_dir).glob("*.mat"))

print(f"Organizing {len(mat_files)} files...")
print("=" * 50)

controls_count = 0
patients_count = 0

for mat_file in mat_files:
    try:
        # Load to check diagnosis
        mat_data = loadmat(mat_file)
        diagnosis = mat_data['diagnosis'][0]  # Extract diagnosis string
        
        # Move to appropriate folder
        if diagnosis == 'Control':
            shutil.move(str(mat_file), os.path.join(controls_dir, mat_file.name))
            print(f"✓ Control: {mat_file.name}")
            controls_count += 1
        elif diagnosis == 'Patient':
            shutil.move(str(mat_file), os.path.join(patients_dir, mat_file.name))
            print(f"✓ Patient: {mat_file.name}")
            patients_count += 1
            
    except Exception as e:
        print(f"✗ ERROR: {mat_file.name} - {e}")

print("=" * 50)
print(f"Moved {controls_count} controls to: {controls_dir}/")
print(f"Moved {patients_count} patients to: {patients_dir}/")
```

```python
"""
NEW CELL FOR COBRE NOTEBOOK: HMM Network Construction
======================================================
Add this as a new cell in your COBRE_data_preparation.ipynb
after the existing cells that create control_data.

This cell constructs HMM-based networks (K=5 states) for COBRE subjects.
"""

# ============================================================================
# HMM NETWORK CONSTRUCTION FOR COBRE
# ============================================================================
from scipy.io import savemat
import os
import numpy as np
from hmmlearn.hmm import GaussianHMM
import warnings
warnings.filterwarnings('ignore')
from hmmlearn.hmm import GaussianHMM
import warnings
warnings.filterwarnings('ignore')


def fit_group_hmm_cobre(control_data, n_states=5, n_iter=100):
    """
    Fit group-level HMM on all COBRE subjects.
    
    Parameters:
    -----------
    control_data : list
        List of subject dictionaries with 'time_series' key
    n_states : int
        Number of brain states (K)
    n_iter : int
        Number of HMM iterations
    
    Returns:
    --------
    hmm : GaussianHMM
        Fitted HMM model
    """
    print("=" * 70)
    print(f"FITTING GROUP-LEVEL HMM (K={n_states} states)")
    print("=" * 70)
    
    # Stack all time series
    all_timeseries = []
    lengths = []
    
    for subj in control_data:
        ts = subj['time_series']  # (timepoints, 116)
        all_timeseries.append(ts)
        lengths.append(len(ts))
    
    # Concatenate all time series
    X = np.vstack(all_timeseries)  # (total_timepoints, 116)
    
    print(f"Total subjects: {len(control_data)}")
    print(f"Total timepoints: {X.shape[0]}")
    print(f"Number of ROIs: {X.shape[1]}")
    print(f"Number of states: {n_states}")
    print(f"HMM iterations: {n_iter}")
    print("\nFitting HMM (this may take a few minutes)...")
    
    # Fit HMM
    hmm = GaussianHMM(
        n_components=n_states,
        covariance_type='full',
        n_iter=n_iter,
        random_state=42,
        verbose=False
    )
    
    hmm.fit(X, lengths)
    
    print("✓ HMM fitting complete!")
    print(f"  Final log-likelihood: {hmm.score(X, lengths):.2f}")
    print("=" * 70)
    
    return hmm


def extract_state_connectivity_cobre(subject_timeseries, hmm, n_states=5, threshold=0.4):
    """
    Extract state-specific connectivity matrices for one subject.
    
    Parameters:
    -----------
    subject_timeseries : np.ndarray
        Time series shape (timepoints, 116)
    hmm : GaussianHMM
        Fitted HMM model
    n_states : int
        Number of states
    threshold : float
        Correlation threshold
    
    Returns:
    --------
    state_networks : np.ndarray
        Shape (n_states, 116, 116) - connectivity per state
    state_counts : np.ndarray
        Number of timepoints per state
    """
    # Predict state sequence
    state_sequence = hmm.predict(subject_timeseries)
    
    state_networks = []
    state_counts = []
    
    for k in range(n_states):
        # Get timepoints belonging to state k
        state_mask = (state_sequence == k)
        state_timepoints = subject_timeseries[state_mask, :]
        
        n_timepoints = np.sum(state_mask)
        state_counts.append(n_timepoints)
        
        # Compute correlation if enough timepoints
        if n_timepoints >= 10:  # Minimum 10 timepoints for stable correlation
            corr_matrix = np.corrcoef(state_timepoints.T)  # (116, 116)
            corr_matrix = np.nan_to_num(corr_matrix, nan=0.0)
        else:
            # Fallback: use whole time series if state has too few timepoints
            corr_matrix = np.corrcoef(subject_timeseries.T)
            corr_matrix = np.nan_to_num(corr_matrix, nan=0.0)
        
        # Apply threshold
        edge_mask = (np.abs(corr_matrix) >= threshold)
        np.fill_diagonal(edge_mask, False)
        
        weighted_adjacency = corr_matrix.copy()
        weighted_adjacency[~edge_mask] = 0.0
        np.fill_diagonal(weighted_adjacency, 0)
        
        state_networks.append(weighted_adjacency)
    
    return np.array(state_networks), np.array(state_counts)


def process_all_cobre_hmm(control_data, hmm, n_states=5, threshold=0.4, 
                           output_dir="COBRE_Networks_HMM"):
    """
    Process all COBRE subjects and extract HMM state-specific networks.
    
    Parameters:
    -----------
    control_data : list
        List of subject dictionaries
    hmm : GaussianHMM
        Fitted HMM model
    n_states : int
        Number of states
    threshold : float
        Correlation threshold
    output_dir : str
        Output directory for .mat files
    """
    os.makedirs(output_dir, exist_ok=True)
    
    print("\n" + "=" * 70)
    print("EXTRACTING STATE-SPECIFIC NETWORKS")
    print("=" * 70)
    print(f"Processing {len(control_data)} subjects")
    print(f"Number of states: {n_states}")
    print(f"Threshold: {threshold}")
    print(f"Output directory: {output_dir}")
    print("=" * 70)
    
    success_count = 0
    fail_count = 0
    
    for i, subject in enumerate(control_data):
        subject_id = subject['subject_id']
        
        try:
            # Extract state-specific connectivity
            state_networks, state_counts = extract_state_connectivity_cobre(
                subject['time_series'],
                hmm,
                n_states=n_states,
                threshold=threshold
            )
            
            # Save to .mat file
            matlab_filename = f"{subject_id}_hmm.mat"
            matlab_filepath = os.path.join(output_dir, matlab_filename)
            
            matlab_data = {
                'weighted_adjacency_matrices': state_networks,  # (K, 116, 116)
                'state_counts': state_counts,  # (K,) - timepoints per state
                'n_states': n_states,
                'threshold': threshold,
                'age': subject['age'],
                'gender': subject['gender'],
                'diagnosis': subject['diagnosis'],
                'TR': 2.0,
                'method': 'HMM'
            }
            
            savemat(matlab_filepath, matlab_data)
            
            success_count += 1
            state_dist = ", ".join([f"S{k}:{state_counts[k]}" for k in range(n_states)])
            print(f"✓ [{i+1}/{len(control_data)}]: {subject_id} - {subject['diagnosis']} "
                  f"{subject['gender']}{subject['age']} ({state_dist} timepoints)")
            
        except Exception as e:
            fail_count += 1
            print(f"✗ [{i+1}/{len(control_data)}]: {subject_id} - {e}")
    
    print("=" * 70)
    print(f"COMPLETE: {success_count} successful, {fail_count} failed")
    print(f"Networks saved to: {output_dir}/")
    print("=" * 70)
    
    return success_count, fail_count


# ============================================================================
# RUN HMM PROCESSING FOR COBRE
# ============================================================================

print("=" * 70)
print("HMM NETWORK CONSTRUCTION - COBRE DATASET")
print("=" * 70)
print(f"Configuration:")
print(f"  - Number of states (K): 5")
print(f"  - Threshold: 0.4")
print(f"  - TR: 2.0s")
print(f"  - Dataset: All COBRE subjects (patients + controls)")
print("=" * 70)

# Check control_data exists
if 'control_data' not in locals():
    print("✗ ERROR: control_data not found!")
    print("  Please run the previous cells to load COBRE data first.")
else:
    # Step 1: Fit group-level HMM
    print("\n" + "="*70)
    print("STEP 1: FITTING HMM")
    print("="*70)
    
    hmm_cobre = fit_group_hmm_cobre(control_data, n_states=5, n_iter=100)
    
    # Step 2: Extract state-specific networks
    print("\n" + "="*70)
    print("STEP 2: EXTRACTING NETWORKS")
    print("="*70)
    
    success, fail = process_all_cobre_hmm(
        control_data,
        hmm_cobre,
        n_states=5,
        threshold=0.4,
        output_dir="COBRE_Networks_HMM"
    )
    
    print("\n" + "="*70)
    print("✓ HMM NETWORK CONSTRUCTION COMPLETE!")
    print("="*70)
    print(f"Processed {len(control_data)} subjects")
    print(f"Output: COBRE_Networks_HMM/")
    print(f"Network shape per subject: (5, 116, 116)")
    
    # Print diagnosis breakdown
    controls = [s for s in control_data if s['diagnosis'] == 'Control']
    patients = [s for s in control_data if s['diagnosis'] == 'Patient']
    print(f"\nDiagnosis breakdown:")
    print(f"  Controls: {len(controls)}")
    print(f"  Patients: {len(patients)}")
    print("="*70)
```
