```python
import pandas as pd
import os
import shutil

# Paths - UPDATE THESE IF DIFFERENT
csv_path = '/Users/noha/Desktop/W2025/Data/query_results.csv'
files_dir = '/Users/noha/Desktop/W2025/Data/Schizophrenia_files'
backup_dir = '/Users/noha/Desktop/W2025/Data/Schizophrenia_files_duplicates'

# Load CSV
df = pd.read_csv(csv_path)

# Get list of files we have
our_files = [f.replace('_CM.nii', '') for f in os.listdir(files_dir) if f.endswith('_CM.nii')]
our_datauris = [int(f) for f in our_files if f.replace('.', '').isdigit()]

print("="*70)
print("ANALYZING YOUR DATA")
print("="*70)
print(f"Total .nii files found: {len(our_files)}")
print(f"Total datauris: {len(our_datauris)}")

# Filter CSV to only scans we have
df_ours = df[df['Datauri'].isin(our_datauris)].copy()

print(f"\nScans matched in CSV: {len(df_ours)}")
print(f"Unique subjects: {df_ours['Subject ID'].nunique()}")

# Find subjects with multiple scans
subject_counts = df_ours['Subject ID'].value_counts()
duplicates = subject_counts[subject_counts > 1]

print(f"\nSubjects with multiple scans: {len(duplicates)}")

if len(duplicates) > 0:
    print("\n" + "="*70)
    print("SUBJECTS WITH MULTIPLE SESSIONS:")
    print("="*70)
    for subject_id in duplicates.index:
        subject_scans = df_ours[df_ours['Subject ID'] == subject_id][['Subject ID', 'Datauri', 'Img Date', 'Age', 'Dx']]
        print(f"\n{subject_id}:")
        print(subject_scans.to_string(index=False))

# Strategy: Keep the FIRST scan (earliest date) for each subject
keep_datauris = []
remove_datauris = []

for subject_id in df_ours['Subject ID'].unique():
    subject_scans = df_ours[df_ours['Subject ID'] == subject_id].sort_values('Img Date')
    # Keep first scan
    keep_datauris.append(int(subject_scans.iloc[0]['Datauri']))
    # Mark rest for removal
    if len(subject_scans) > 1:
        for datauri in subject_scans.iloc[1:]['Datauri']:
            remove_datauris.append(int(datauri))

print("\n" + "="*70)
print("SELECTION SUMMARY")
print("="*70)
print(f"Scans to KEEP: {len(keep_datauris)}")
print(f"Scans to REMOVE: {len(remove_datauris)}")

# Create backup directory for removed files
if len(remove_datauris) > 0:
    os.makedirs(backup_dir, exist_ok=True)
    
    print("\n" + "="*70)
    print("MOVING DUPLICATE SCANS TO BACKUP")
    print("="*70)
    
    for datauri in remove_datauris:
        source = os.path.join(files_dir, f"{datauri}_CM.nii")
        dest = os.path.join(backup_dir, f"{datauri}_CM.nii")
        if os.path.exists(source):
            shutil.move(source, dest)
            print(f"Moved: {datauri}_CM.nii")
    
    print(f"\nDuplicate scans backed up to: {backup_dir}")

# Save final subject list
final_subjects = df_ours[df_ours['Datauri'].isin(keep_datauris)]['Subject ID'].unique()
with open(os.path.join(files_dir, 'final_subjects_list.txt'), 'w') as f:
    for subj in sorted(final_subjects):
        f.write(f"{subj}\n")

# Count remaining files
remaining_files = [f for f in os.listdir(files_dir) if f.endswith('_CM.nii')]

print("\n" + "="*70)
print("FINAL RESULT")
print("="*70)
print(f"Remaining .nii files: {len(remaining_files)}")
print(f"Unique subjects: {len(final_subjects)}")

# Save diagnosis breakdown
print("\nDiagnosis breakdown of final dataset:")
diagnosis_counts = df_ours[df_ours['Datauri'].isin(keep_datauris)]['Dx'].value_counts()
print(diagnosis_counts)

# Save subject info
final_df = df_ours[df_ours['Datauri'].isin(keep_datauris)][['Subject ID', 'Datauri', 'Age', 'Sex', 'Dx']]
final_df.to_csv(os.path.join(files_dir, 'final_subjects_info.csv'), index=False)
print(f"\nSubject information saved to: final_subjects_info.csv")

print("\n" + "="*70)
print("DONE! Your data is ready for analysis.")
print("="*70)
```
