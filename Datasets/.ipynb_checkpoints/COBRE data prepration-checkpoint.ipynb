```python
import tarfile
import os

print('Attempting to extract files with error recovery...')
extracted_count = 0
failed_count = 0

try:
    with tarfile.open('raw_rfMRI.tar', 'r', ignore_zeros=True, errorlevel=0) as tar:
        for member in tar:
            try:
                if member.isfile() and member.name.endswith('.nii.gz'):
                    tar.extract(member, path='.')
                    print(f'✓ Extracted: {member.name}')
                    extracted_count += 1
                elif member.isdir():
                    tar.extract(member, path='.')
                    print(f'✓ Created directory: {member.name}')
            except Exception as e:
                print(f'✗ Failed: {member.name} - {e}')
                failed_count += 1
                continue
                
    print(f'\\nSummary:')
    print(f'Successfully extracted: {extracted_count} files')
    print(f'Failed extractions: {failed_count} files')
    
except Exception as e:
    print(f'Error opening archive: {e}')
```

```python
import nibabel as nib
import numpy as np
import os
import glob
from nilearn.maskers import NiftiLabelsMasker

# Point to the AAL1 atlas files (116 regions)
extract_dir = "/Users/noha/Downloads/aal_for_SPM8"  # Based on your screenshot
aal_map = os.path.join(extract_dir, "ROI_MNI_V4.nii")
aal_labels_file = os.path.join(extract_dir, "ROI_MNI_V4.txt")

# Read labels
aal_labels = []
with open(aal_labels_file, 'r') as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 2:
            aal_labels.append(parts[1])

# Load the actual atlas data
atlas_img = nib.load(aal_map)
print(f"Atlas shape: {atlas_img.shape}")
print(f"Number of region labels: {len(aal_labels)}")
print(f"Unique values in atlas: {len(np.unique(atlas_img.get_fdata()))}")

# Verify it's 116 regions
assert len(aal_labels) == 116, f"Expected 116 regions, got {len(aal_labels)}"
print("✓ Confirmed: This is the 116-region AAL atlas!")

# Create masker for 116 regions (excluding background)
masker = NiftiLabelsMasker(
    labels_img=aal_map,
    standardize=True,
    detrend=True,
    verbose=1
)

# Use the raw_rfMRI folder in the same directory as the notebook
data_dir = "raw_rfMRI"

# Get all .nii.gz files
swedish_files = glob.glob(os.path.join(data_dir, "*.nii.gz"))
print(f"\nFound {len(swedish_files)} control subjects")

# Process each control subject
control_data = []

for i, subject_file in enumerate(swedish_files):
    print(f"\nProcessing subject {i+1}/{len(swedish_files)}: {os.path.basename(subject_file)}")
    
    try:
        # Load 4D fMRI data
        img = nib.load(subject_file)
        print(f"Subject fMRI shape: {img.shape}")
        
        # Extract time series for 116 AAL regions
        time_series = masker.fit_transform(img)
        print(f"Time series shape: {time_series.shape}")  # Should be (150, 116)
        
        # Verify exactly 116 regions
        assert time_series.shape[1] == 116, f"Expected 116 regions, got {time_series.shape[1]}"
        
        # Extract age and gender from filename
        filename = os.path.basename(subject_file)
        age = int(filename[1:3])
        gender = filename[0]
        
        # Store the time series
        control_data.append({
            'subject_id': filename,
            'time_series': time_series,
            'age': age,
            'gender': gender
        })
        
        print(f"✓ Successfully processed {gender}{age}")
        
    except Exception as e:
        print(f"✗ Error processing {subject_file}: {e}")
        continue

print(f"\n{'='*50}")
print(f"✓ Successfully processed {len(control_data)} control subjects")
if control_data:
    print(f"Age range: {min([s['age'] for s in control_data])} - {max([s['age'] for s in control_data])}")
```

```python
# Check the results
print(f"Total subjects processed: {len(control_data)}")
print(f"\nAge distribution:")
ages = [s['age'] for s in control_data]
print(f"  Range: {min(ages)} - {max(ages)}")
print(f"  Mean: {np.mean(ages):.1f}")

print(f"\nGender distribution:")
genders = [s['gender'] for s in control_data]
print(f"  Male: {genders.count('m')}")
print(f"  Female: {genders.count('f')}")

print(f"\nTime series info:")
print(f"  Shape of each time series: {control_data[0]['time_series'].shape}")
print(f"  Number of brain regions: {control_data[0]['time_series'].shape[1]}")

print(f"\nFirst 3 subjects:")
for i in range(3):
    s = control_data[i]
    print(f"  {i+1}. {s['subject_id']}: {s['gender']}{s['age']}, shape={s['time_series'].shape}")
```

```python
import pandas as pd
import numpy as np
import networkx as nx
from pathlib import Path
from scipy.io import savemat
import os 

def calculate_time_windows(n_timepoints, TR=2.5, window_duration=30):
    """Calculate window size and number of windows"""
    window_size = int(window_duration / TR)
    n_windows = n_timepoints // window_size
    
    return window_size, n_windows

def create_time_windows(data, window_size, n_windows):
    """Split time series data into windows"""
    windows = []
    for i in range(n_windows):
        start_idx = i * window_size
        end_idx = start_idx + window_size
        window_data = data[start_idx:end_idx, :]  # Note: transposed from original
        windows.append(window_data)
    
    return windows

def calculate_correlation_matrix(window_data):
    """Calculate correlation matrix for a time window"""
    # window_data shape: (timepoints, regions)
    correlation_matrix = np.corrcoef(window_data.T)  # Transpose to get region x region
    correlation_matrix = np.nan_to_num(correlation_matrix, nan=0.0)
    
    return correlation_matrix

def create_weighted_network_layer(correlation_matrix, threshold):
    """Create weighted network from correlation matrix"""
    edge_mask = (np.abs(correlation_matrix) >= threshold)
    np.fill_diagonal(edge_mask, False)
    
    weighted_adjacency = correlation_matrix.copy()
    weighted_adjacency[~edge_mask] = 0.0
    np.fill_diagonal(weighted_adjacency, 0)
    
    G = nx.from_numpy_array(weighted_adjacency)
    
    for i, j in G.edges():
        G[i][j]['correlation'] = correlation_matrix[i, j]
        G[i][j]['abs_correlation'] = abs(correlation_matrix[i, j])
        
    return G, weighted_adjacency

def construct_multilayer_network_control(subject_data, window_duration=30, 
                                        absolute_threshold=0.4, TR=2.0):
    """
    Construct multilayer network for control subject
    
    Parameters:
    -----------
    subject_data : dict
        Dictionary containing 'subject_id', 'time_series', 'age', 'gender'
    window_duration : int
        Window duration in seconds (default: 30)
    absolute_threshold : float
        Correlation threshold (default: 0.4)
    TR : float
        Repetition time in seconds (default: 2.0)
    """
    
    timeseries_data = subject_data['time_series']  # Shape: (timepoints, 116)
    subject_id = subject_data['subject_id'].replace('.nii.gz', '')
    
    n_timepoints, n_regions = timeseries_data.shape
    
    # Calculate windows
    window_size, n_windows = calculate_time_windows(n_timepoints, TR, window_duration)
    
    if n_windows == 0:
        raise ValueError(f"No complete windows possible. Need at least {window_size} timepoints, have {n_timepoints}")
    
    # Create time windows
    windows = create_time_windows(timeseries_data, window_size, n_windows)
    
    # Calculate correlation matrices for each window
    correlation_matrices = []
    for window_data in windows:
        corr_matrix = calculate_correlation_matrix(window_data)
        correlation_matrices.append(corr_matrix)
    
    # Create network layers
    layers = []
    weighted_adjacency_matrices = []
    
    for corr_matrix in correlation_matrices:
        layer_graph, weighted_adj_matrix = create_weighted_network_layer(
            corr_matrix, absolute_threshold
        )
        layers.append(layer_graph)
        weighted_adjacency_matrices.append(weighted_adj_matrix)
    
    results = {
        'layers': layers,
        'weighted_adjacency_matrices': weighted_adjacency_matrices,
        'subject_id': subject_id,
        'age': subject_data['age'],
        'gender': subject_data['gender'],
        'n_windows': n_windows,
        'n_regions': n_regions
    }
    
    return results

def process_all_controls(control_data, window_duration=30, absolute_threshold=0.4, 
                        TR=2.0, output_dir="Networks_controls"):
    """
    Process all control subjects and save networks
    
    Parameters:
    -----------
    control_data : list
        List of dictionaries with subject data
    window_duration : int
        Window duration in seconds
    absolute_threshold : float
        Correlation threshold
    TR : float
        Repetition time in seconds
    output_dir : str
        Directory to save MATLAB files
    """
    
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    
    all_results = []
    success_count = 0
    fail_count = 0
    
    print(f"Processing {len(control_data)} control subjects")
    print(f"Parameters: TR={TR}s, Window={window_duration}s, Threshold={absolute_threshold}")
    print("="*60)
    
    for i, subject_data in enumerate(control_data):
        subject_id = subject_data['subject_id'].replace('.nii.gz', '')
        
        try:
            # Construct multilayer network
            results = construct_multilayer_network_control(
                subject_data,
                window_duration=window_duration,
                absolute_threshold=absolute_threshold,
                TR=TR
            )
            
            # Save to MATLAB format
            matlab_filename = f"{subject_id}.mat"
            matlab_filepath = os.path.join(output_dir, matlab_filename)
            
            matlab_data = {
                'weighted_adjacency_matrices': np.array(results['weighted_adjacency_matrices']),
                'age': results['age'],
                'gender': results['gender'],
                'n_windows': results['n_windows'],
                'n_regions': results['n_regions']
            }
            
            savemat(matlab_filepath, matlab_data)
            
            all_results.append(results)
            success_count += 1
            print(f"✓ SUCCESS [{i+1}/{len(control_data)}]: {subject_id} - "
                  f"{results['gender']}{results['age']}, {results['n_windows']} windows")
            
        except Exception as e:
            fail_count += 1
            print(f"✗ ERROR [{i+1}/{len(control_data)}]: {subject_id} - {e}")
    
    print("="*60)
    print(f"COMPLETE: {success_count} successful, {fail_count} failed")
    print(f"MATLAB files saved to: {output_dir}/")
    
    # Create summary CSV
    summary_data = []
    for result in all_results:
        summary_data.append({
            'subject_id': result['subject_id'],
            'age': result['age'],
            'gender': result['gender'],
            'n_windows': result['n_windows'],
            'n_regions': result['n_regions']
        })
    
    if summary_data:
        summary_df = pd.DataFrame(summary_data)
        summary_path = os.path.join(output_dir, 'processing_summary.csv')
        summary_df.to_csv(summary_path, index=False)
        print(f"Summary saved to: {summary_path}")
    
    return all_results

# Run the processing
if __name__ == "__main__":
    # Process all control subjects
    # Assuming control_data is already loaded in your notebook
    results = process_all_controls(
        control_data,
        window_duration=30,
        absolute_threshold=0.4,
        TR=2.5,  # Adjust this if you know the actual TR
        output_dir="Networks_controls"
    )
    
    print("\nProcessing complete!")
```
