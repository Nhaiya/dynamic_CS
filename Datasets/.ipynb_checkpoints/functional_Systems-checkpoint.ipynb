```python
import pandas as pd
import numpy as np
from nilearn import image, datasets
from scipy import stats
from nibabel import load as nib_load

# Correct path to AAL atlas
aal_path = '/Users/noha/Downloads/aal/ROI_MNI_V4.nii'

print("Loading AAL atlas...")
aal_img = nib_load(aal_path)

# Load your AAL labels
print("Loading AAL labels...")
aal_labels = pd.read_csv('/Users/noha/Desktop/W2025/Data/Multiple_Sclerosis/AAL_labels.csv')

# Download Yeo atlas
print("Downloading Yeo atlas...")
yeo_atlas = datasets.fetch_atlas_yeo_2011()
yeo_7_img = image.load_img(yeo_atlas['thick_7'])

print("Resampling Yeo atlas to AAL space...")
yeo_resampled = image.resample_to_img(yeo_7_img, aal_img, interpolation='nearest', 
                                       force_resample=True, copy_header=True)

print("Extracting atlas data...")
aal_data = aal_img.get_fdata()
yeo_data = yeo_resampled.get_fdata()

# Get unique AAL values from the actual data
unique_aal_values = np.unique(aal_data)
unique_aal_values = unique_aal_values[unique_aal_values > 0]  # Remove background

# Map using actual AAL atlas values
print("Mapping AAL regions to Yeo networks...")
aal_value_to_yeo = {}

for aal_value in unique_aal_values:
    aal_mask = (aal_data == aal_value)
    yeo_values = yeo_data[aal_mask]
    yeo_nonzero = yeo_values[yeo_values > 0]
    
    if len(yeo_nonzero) > 0:
        most_common = stats.mode(yeo_nonzero, keepdims=True)[0][0]
        aal_value_to_yeo[int(aal_value)] = int(most_common)
    else:
        aal_value_to_yeo[int(aal_value)] = 0

# Convert AAL atlas values to ROI numbers
# Pattern: 2001->1, 2002->2, 2101->3, 2102->4, etc.
# The mapping appears to be sequential based on the atlas values
aal_value_to_roi = {}
sorted_values = sorted(unique_aal_values)
for idx, aal_value in enumerate(sorted_values, start=1):
    aal_value_to_roi[int(aal_value)] = idx

print(f"\nMapped {len(aal_value_to_yeo)} regions")
print(f"Sample AAL value to ROI: {dict(list(aal_value_to_roi.items())[:10])}")

# Create final mapping using ROI numbers
aal_to_yeo = {}
for aal_value, yeo_network in aal_value_to_yeo.items():
    roi_number = aal_value_to_roi[aal_value]
    aal_to_yeo[roi_number] = yeo_network

# Define Yeo network names
yeo_network_names = {
    0: 'Background',
    1: 'Visual',
    2: 'Somatomotor',
    3: 'Dorsal Attention',
    4: 'Ventral Attention',
    5: 'Limbic',
    6: 'Frontoparietal',
    7: 'Default Mode'
}

# Create mapping dataframe
mapping_df = pd.DataFrame({
    'AAL_ID': list(aal_to_yeo.keys()),
    'Yeo_Network_ID': list(aal_to_yeo.values())
})

mapping_df['Yeo_Network_Name'] = mapping_df['Yeo_Network_ID'].map(yeo_network_names)

# Merge with AAL labels
final_mapping = pd.merge(aal_labels, mapping_df, left_on='ROI number', right_on='AAL_ID', how='left')
final_mapping = final_mapping[['ROI number', 'ROI name', 'Yeo_Network_ID', 'Yeo_Network_Name']]

# Save the output
output_path = '/Users/noha/Desktop/AAL_to_Yeo7_mapping.csv'
final_mapping.to_csv(output_path, index=False)

print(f"\n✓ Mapping saved successfully to {output_path}!")
print(f"✓ Total regions mapped: {len(final_mapping)}")
print("\nNetwork distribution:")
print(final_mapping['Yeo_Network_Name'].value_counts().sort_index())
print("\nFirst 20 rows:")
print(final_mapping.head(20))
print("\nLast 10 rows:")
print(final_mapping.tail(10))
```

```python
import pandas as pd

def map_rois_to_networks(roi_list, mapping_file='//Users/noha/Desktop/W2025/Data/AAL_to_Yeo7_mapping.csv'):
 
    # Load the mapping
    aal_yeo_mapping = pd.read_csv(mapping_file)
    
    # Filter for requested ROIs
    node_networks = aal_yeo_mapping[aal_yeo_mapping['ROI number'].isin(roi_list)]
    node_networks = node_networks.sort_values('ROI number')
    
    # Print results
    print("="*80)
    print(f"Mapping {len(roi_list)} ROIs to Yeo Networks")
    print("="*80)
    print(node_networks.to_string(index=False))
    
    print("\n" + "="*80)
    print("Network Summary")
    print("="*80)
    network_counts = node_networks['Yeo_Network_Name'].value_counts()
    for network, count in network_counts.items():
        percentage = (count / len(roi_list)) * 100
        print(f"{network}: {count} ROIs ({percentage:.1f}%)")
    
    print("\n" + "="*80)
    print("Grouped by Network")
    print("="*80)
    for network in sorted(node_networks['Yeo_Network_Name'].unique()):
        network_nodes = node_networks[node_networks['Yeo_Network_Name'] == network]
        roi_list_net = network_nodes['ROI number'].tolist()
        print(f"\n{network}: {roi_list_net}")
        for _, row in network_nodes.iterrows():
            print(f"  ROI {row['ROI number']:3d}: {row['ROI name']}")
    
    return node_networks

# Example usage
nodes = [  2, 46, 47, 66, 83, 84, 91, 98, 99, 102, 103, 104, 112 ]
result = map_rois_to_networks(nodes)

# You can also access the result as a dataframe
print("\n" + "="*80)
print("As DataFrame:")
print(result)
```

```python
#!/usr/bin/env python3
"""
Extend Yeo 7-network mapping to subcortical structures
Jupyter notebook version - no command line arguments

Just set the file paths and run create_complete_aal_network_mapping()
"""

import os
import requests
import nibabel as nib
import numpy as np
import pandas as pd
from scipy import ndimage
import warnings
warnings.filterwarnings('ignore')


def download_ji_subcortical_network_atlas(output_dir='.'):
    """
    Download Ji et al. 2019 subcortical atlas with network assignments
    """
    urls = [
        "https://www.colelab.org/data/Ji2019_SubcorticalNetworkParcels.nii.gz",
        "https://neurovault.org/media/images/4498/ColeAnticevicNetPartition_v1_SubcortexOnly.nii.gz",
    ]
    
    for url in urls:
        try:
            print(f"Downloading Ji atlas from: {url}")
            response = requests.get(url, timeout=60, verify=False)
            
            if response.status_code == 200:
                fname = os.path.join(output_dir, 'Ji2019_Subcortical_Networks.nii.gz')
                with open(fname, 'wb') as f:
                    f.write(response.content)
                
                img = nib.load(fname)
                data = img.get_fdata()
                
                print(f"✓ Downloaded: {fname}")
                print(f"  Shape: {data.shape}")
                return fname
                
        except Exception as e:
            print(f"✗ Failed: {e}")
    
    return None


def download_buckner_cerebellum_networks(output_dir='.'):
    """
    Download Buckner cerebellum atlas with Yeo network assignments
    """
    urls = [
        "https://github.com/DiedrichsenLab/cerebellar_atlases/raw/master/Buckner_2011/atl-Buckner7_space-MNI_dseg.nii",
        "https://github.com/ThomasYeoLab/CBIG/raw/master/stable_projects/brain_parcellation/Yeo2011_fcMRI_clustering/1000subjects_reference/Yeo_JNeurophysiol11_MNI152/Buckner2011_7Networks_MNI152_FreeSurferConformed1mm.nii.gz",
    ]
    
    for url in urls:
        try:
            print(f"Downloading Buckner atlas from: {url}")
            response = requests.get(url, timeout=60, verify=False)
            
            if response.status_code == 200:
                fname = os.path.join(output_dir, 'Buckner_Cerebellum_7Networks.nii.gz')
                with open(fname, 'wb') as f:
                    f.write(response.content)
                
                print(f"✓ Downloaded: {fname}")
                return fname
                
        except Exception as e:
            print(f"✗ Failed: {e}")
    
    return None


def assign_networks_by_overlap(aal_atlas_path, network_atlas_path, atlas_type='yeo'):
    """
    Assign networks to AAL regions based on spatial overlap
    """
    network_names = {
        0: 'Background',
        1: 'Visual',
        2: 'Somatomotor', 
        3: 'Dorsal Attention',
        4: 'Ventral Attention',
        5: 'Limbic',
        6: 'Frontoparietal',
        7: 'Default Mode'
    }
    
    # Load atlases
    print(f"  Loading AAL atlas...")
    aal_img = nib.load(aal_atlas_path)
    aal_data = aal_img.get_fdata()
    
    print(f"  Loading {atlas_type} network atlas...")
    net_img = nib.load(network_atlas_path)
    net_data = net_img.get_fdata()
    
    # Resample if needed
    if aal_data.shape != net_data.shape:
        print(f"  Resampling {atlas_type} atlas to AAL space...")
        try:
            from nilearn.image import resample_img
            net_img = resample_img(
                net_img,
                target_affine=aal_img.affine,
                target_shape=aal_img.shape,
                interpolation='nearest'
            )
            net_data = net_img.get_fdata()
        except ImportError:
            print("  ERROR: nilearn required for resampling")
            print("  Install: pip install nilearn")
            return {}
    
    # Compute overlap
    assignments = {}
    aal_rois = np.unique(aal_data)
    aal_rois = aal_rois[aal_rois > 0]
    
    print(f"  Computing overlap for {len(aal_rois)} regions...")
    
    for roi in aal_rois:
        roi_mask = (aal_data == roi)
        net_values = net_data[roi_mask]
        
        total_voxels = len(net_values)
        net_values_nonzero = net_values[net_values > 0]
        
        if len(net_values_nonzero) == 0:
            assignments[roi] = {
                'network': 'Background',
                'network_id': 0,
                'coverage': 0.0
            }
            continue
        
        # Most common network
        unique_nets, counts = np.unique(net_values_nonzero, return_counts=True)
        dominant_net_id = unique_nets[np.argmax(counts)]
        dominant_count = counts[np.argmax(counts)]
        
        coverage = (dominant_count / total_voxels) * 100
        network_name = network_names.get(int(dominant_net_id), 'Unknown')
        
        assignments[roi] = {
            'network': network_name,
            'network_id': int(dominant_net_id),
            'coverage': coverage
        }
    
    return assignments


def create_complete_aal_network_mapping(
    aal_atlas_path,
    yeo_atlas_path,
    aal_labels_csv,
    output_csv,
    ji_atlas_path=None,
    buckner_atlas_path=None,
    download_atlases=True,
    min_coverage=5.0
):
    """
    Create complete AAL-to-network mapping
    
    Example usage:
        df = create_complete_aal_network_mapping(
            aal_atlas_path='/path/to/AAL.nii',
            yeo_atlas_path='/path/to/Yeo_7Networks.nii',
            aal_labels_csv='/path/to/AAL_labels.csv',
            output_csv='AAL_to_Networks.csv',
            download_atlases=True
        )
    """
    print("="*70)
    print("EXTENDING YEO NETWORKS TO SUBCORTEX AND CEREBELLUM")
    print("="*70)
    
    # Download if needed
    if download_atlases:
        print("\n1. Attempting to download network atlases...")
        if ji_atlas_path is None:
            ji_atlas_path = download_ji_subcortical_network_atlas()
        if buckner_atlas_path is None:
            buckner_atlas_path = download_buckner_cerebellum_networks()
    
    # Load AAL labels
    print("\n2. Loading AAL labels...")
    aal_labels = pd.read_csv(aal_labels_csv)
    print(f"   {len(aal_labels)} regions")
    
    # Yeo cortical
    print("\n3. Computing Yeo cortical assignments...")
    yeo_assignments = assign_networks_by_overlap(aal_atlas_path, yeo_atlas_path, 'yeo')
    print(f"   Done: {len(yeo_assignments)} regions")
    
    # Ji subcortical
    ji_assignments = {}
    if ji_atlas_path and os.path.exists(ji_atlas_path):
        print("\n4. Computing Ji subcortical assignments...")
        ji_assignments = assign_networks_by_overlap(aal_atlas_path, ji_atlas_path, 'ji')
        print(f"   Done: {len(ji_assignments)} regions")
    else:
        print("\n4. Ji atlas not available - using Yeo only for subcortex")
    
    # Buckner cerebellum
    buckner_assignments = {}
    if buckner_atlas_path and os.path.exists(buckner_atlas_path):
        print("\n5. Computing Buckner cerebellar assignments...")
        buckner_assignments = assign_networks_by_overlap(aal_atlas_path, buckner_atlas_path, 'buckner')
        print(f"   Done: {len(buckner_assignments)} regions")
    else:
        print("\n5. Buckner atlas not available - using Yeo only for cerebellum")
    
    # Combine
    print("\n6. Combining assignments...")
    final_mapping = []
    
    for idx, row in aal_labels.iterrows():
        roi_num = row['ROI']
        region_name = row['Label']
        
        # Default: Yeo
        if roi_num in yeo_assignments:
            assignment = yeo_assignments[roi_num]
            network = assignment['network']
            coverage = assignment['coverage']
            source = f"Yeo cortical ({coverage:.1f}%)"
        else:
            network = 'Unassigned'
            coverage = 0.0
            source = 'No overlap'
        
        # Override with Ji if better
        if roi_num in ji_assignments and ji_assignments[roi_num]['coverage'] > 0:
            ji_assign = ji_assignments[roi_num]
            network = ji_assign['network']
            coverage = ji_assign['coverage']
            source = f"Ji subcortical ({coverage:.1f}%)"
        
        # Override with Buckner if better
        if roi_num in buckner_assignments and buckner_assignments[roi_num]['coverage'] > 0:
            buckner_assign = buckner_assignments[roi_num]
            network = buckner_assign['network']
            coverage = buckner_assign['coverage']
            source = f"Buckner cerebellum ({coverage:.1f}%)"
        
        # Min coverage threshold
        if coverage < min_coverage and network != 'Unassigned':
            network = 'Background'
            source = f"Low coverage ({coverage:.1f}%)"
        
        final_mapping.append({
            'ROI': roi_num,
            'Region_Name': region_name,
            'Functional_Network': network,
            'Coverage_%': coverage,
            'Data_Source': source
        })
    
    # Save
    df = pd.DataFrame(final_mapping)
    df.to_csv(output_csv, index=False)
    print(f"\n✓ Saved: {output_csv}")
    
    # Summary
    print("\n" + "="*70)
    print("NETWORK DISTRIBUTION")
    print("="*70)
    for network, count in df['Functional_Network'].value_counts().sort_index().items():
        pct = (count / len(df)) * 100
        print(f"  {network:.<30} {count:>3} ({pct:>5.1f}%)")
    
    print(f"\n{'TOTAL':.>36} {len(df):>3}")
    
    # Low coverage warning
    low = df[df['Functional_Network'] == 'Background']
    if len(low) > 0:
        print(f"\n⚠ {len(low)} regions with low coverage")
    
    return df


# ============================================================================
# EXAMPLE USAGE - EDIT THESE PATHS AND RUN
# ============================================================================

if __name__ == "__main__":
    # EDIT THESE PATHS:
    aal_atlas = '/path/to/AAL.nii'
    yeo_atlas = '/path/to/Yeo_7Networks.nii'
    aal_labels = '/Users/noha/Desktop/W2025/Data/Multiple_Sclerosis/AAL_labels.csv'
    output_file = 'AAL_to_Networks_Complete.csv'
    
    # Run the mapping
    df = create_complete_aal_network_mapping(
        aal_atlas_path=aal_atlas,
        yeo_atlas_path=yeo_atlas,
        aal_labels_csv=aal_labels,
        output_csv=output_file,
        download_atlases=True,  # Set False if you already have Ji/Buckner atlases
        min_coverage=5.0
    )
    
    # Display results
    print("\nFirst 20 rows:")
    print(df.head(20))
```
